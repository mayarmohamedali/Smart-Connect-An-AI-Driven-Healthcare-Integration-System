```python
#444 - COMPREHENSIVE DATA CLEANING
print("="*70)
print("DATA CLEANING & PREPROCESSING")
print("="*70)

import re

# 1. DROP LEAKAGE COLUMNS
leakage_cols = ['ID', 'Name', 'checkin date', 'checkout date']
df.drop(columns=[c for c in leakage_cols if c in df.columns], inplace=True)
print(f"✓ Dropped leakage columns")

# 2. CLEAN DIAGNOSIS COLUMN - REMOVE SPECIAL CHARACTERS & ARABIC TEXT
def clean_diagnosis(text):
    """Remove Arabic characters, special symbols, extra spaces"""
    if pd.isna(text):
        return np.nan

    # Convert to string
    text = str(text)

    # Remove Arabic/non-ASCII characters
    text = re.sub(r'[^\x00-\x7F]+', '', text)

    # Remove extra punctuation except comma and hyphen
    text = re.sub(r'[^\w\s,\-]', '', text)

    # Remove extra spaces
    text = ' '.join(text.split())

    # Remove leading/trailing whitespace
    text = text.strip()

    # If empty after cleaning, return NaN
    if not text:
        return np.nan

    return text.lower()

if 'Diagnosis' in df.columns:
    print(f"\n📋 Cleaning Diagnosis column...")
    print(f"Before: {df['Diagnosis'].nunique()} unique diagnoses")
    df['Diagnosis'] = df['Diagnosis'].apply(clean_diagnosis)
    print(f"After: {df['Diagnosis'].nunique()} unique diagnoses")

# 3. CONVERT NUMERIC COLUMNS PROPERLY
numeric_cols = ['CBC-TLC1', 'CBC-TLC2', 'CBC-HB1', 'CBC-HB2',
                'CBC-PLAT1', 'CBC-PLAT2', 'BLOOD-URIA1', 'BLOOD-URIA2',
                'BLOOD-creatinine1', 'BLOOD-creatinine2', 'Age', 'BMI',
                'glucose', 'Systolic_BP(hypertention risk)', 'avg_creatinine',
                'avg_urea', 'avg_hb', 'avg_tlc', 'avg_platelets']

for col in numeric_cols:
    if col in df.columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

print(f"\n✓ Converted {len([c for c in numeric_cols if c in df.columns])} numeric columns")

# 4. DROP ROWS WITH MISSING TARGET (Diagnosis)
if 'Diagnosis' in df.columns:
    before_drop = len(df)
    df = df.dropna(subset=['Diagnosis'])
    after_drop = len(df)
    print(f"✓ Dropped {before_drop - after_drop} rows with missing diagnosis")

# 5. HANDLE MISSING VALUES INTELLIGENTLY
# For medical measurements - use median (more robust than mean)
for col in numeric_cols:
    if col in df.columns and df[col].isnull().sum() > 0:
        median_val = df[col].median()
        df[col].fillna(median_val, inplace=True)

# For categorical - use mode or 'unknown'
cat_cols = ['Gender', 'Smoking_Status', 'Physical_Activity_Level']
for col in cat_cols:
    if col in df.columns and df[col].isnull().sum() > 0:
        mode_val = df[col].mode()[0] if not df[col].mode().empty else 'unknown'
        df[col].fillna(mode_val, inplace=True)

# For binary flags - fill with 0 (no condition)
binary_cols = ['Has_Diabetes', 'Has_Hypertension', 'Has_Kidney_Disease', 'Has_Heart_Disease']
for col in binary_cols:
    if col in df.columns:
        df[col].fillna(0, inplace=True)

# 6. DROP DUPLICATES
before_dup = len(df)
df.drop_duplicates(inplace=True)
after_dup = len(df)
print(f"✓ Dropped {before_dup - after_dup} duplicate rows")

# 7. FINAL CHECK
print(f"\n{'='*70}")
print(f"✅ CLEANING COMPLETE")
print(f"{'='*70}")
print(f"Final shape: {df.shape}")
print(f"Missing values per column:")
missing = df.isnull().sum()
if missing.sum() == 0:
    print("   ✅ NO MISSING VALUES!")
else:
    print(missing[missing > 0])

# Save cleaned version
df_clean = df.copy()
print(f"\n✓ Saved to df_clean")
```

```python
print("="*70)
print("OPTIMIZED DIAGNOSIS GROUPING (TARGET: 10-11 CATEGORIES)")
print("="*70)

def normalize_diagnosis(text):
    """
    Merge diagnoses into 10-11 clear categories
    """
    if pd.isna(text):
        return "Other"

    text = str(text).lower()

    # 1. KIDNEY (Merge ALL kidney issues into ONE category)
    if any(word in text for word in ['kidney', 'renal', 'aki', 'ckd', 'nephritis', 'acute kidney', 'kidney failure']):
        return "Kidney_Disease"

    # 2. PNEUMONIA & LUNG INFECTIONS
    elif any(word in text for word in ['pneumonia', 'lung infection', 'chest infection', 'respiratory infection']):
        return "Pneumonia"

    # 3. SEPSIS & SEVERE INFECTIONS
    elif any(word in text for word in ['sepsis', 'septic']):
        return "Sepsis"

    # 4. UTI
    elif any(word in text for word in ['uti', 'urinary tract']):
        return "UTI"

    # 5. STROKE
    elif any(word in text for word in ['stroke', 'cva', 'cerebrovascular']):
        return "Stroke"

    # 6. HEART DISEASE (Merge all cardiac issues)
    elif any(word in text for word in ['heart', 'cardiac', 'mi ', 'infarct', 'coronary']):
        return "Heart_Disease"

    # 7. DIABETES (Merge all diabetic emergencies)
    elif any(word in text for word in ['diabetes', 'diabetic', 'dka', 'ketoacid', 'hyperglycemia']):
        return "Diabetes_Emergency"

    # 8. ANEMIA
    elif any(word in text for word in ['anemia', 'anaemia']):
        return "Anemia"

    # 9. RESPIRATORY (Non-infection, merge COPD, asthma, etc.)
    elif any(word in text for word in ['copd', 'asthma', 'pulmonary edema', 'respiratory failure', 'bronchitis']):
        return "Respiratory_Disease"

    # 10. GASTROINTESTINAL (Merge all GI issues)
    elif any(word in text for word in ['gastro', 'gi ', 'intestin', 'stomach', 'abdomen', 'liver', 'hepatic', 'cirrhosis']):
        return "Gastrointestinal"

    # 11. Everything else
    else:
        return "Other"

# Apply grouping
df_clean['Diagnosis_Grouped'] = df_clean['Diagnosis'].apply(normalize_diagnosis)

# Show distribution
diagnosis_counts = df_clean['Diagnosis_Grouped'].value_counts()
print("\n📊 Disease distribution:")
for disease, count in diagnosis_counts.items():
    percentage = (count / len(df_clean)) * 100
    print(f"   {disease:30s}: {count:3d} samples ({percentage:5.1f}%)")

print(f"\n✅ Total categories: {len(diagnosis_counts)} (TARGET: 10-11)")
print(f"✅ Total patients: {len(df_clean)}")

# Filter - keep only diseases with ≥8 samples
valid_classes = diagnosis_counts[diagnosis_counts >= 8].index
df_clean = df_clean[df_clean['Diagnosis_Grouped'].isin(valid_classes)]

diagnosis_counts_filtered = df_clean['Diagnosis_Grouped'].value_counts()
print(f"\n✅ Kept {len(diagnosis_counts_filtered)} categories with ≥8 samples")
print(f"✅ Final dataset: {len(df_clean)} patients")

print("\n📊 Final distribution:")
for disease, count in diagnosis_counts_filtered.items():
    percentage = (count / len(df_clean)) * 100
    print(f"   {disease:30s}: {count:3d} samples ({percentage:5.1f}%)")
```

```python
#30 - CREATE EPIDEMIC DATAFRAME (NO TARGET YET!)

print("=" * 70)
print("PREPARING EPIDEMIC DATA")
print("=" * 70)

# Check which diagnosis column exists
if 'Diagnosis' in df.columns:
    diagnosis_col = 'Diagnosis'
elif 'Diagnosis_Grouped' in df.columns:
    diagnosis_col = 'Diagnosis_Grouped'
else:
    print("ERROR: No Diagnosis column found!")
    print(f"Available columns: {df.columns.tolist()}")
    raise ValueError("Diagnosis column not found")

print(f"✓ Using column: {diagnosis_col}")

# Group by month and diagnosis - NO EPIDEMIC COLUMN YET!
epidemic_df = df.groupby(
    ["month", diagnosis_col]
).size().reset_index(name="cases")

# Rename for consistency
epidemic_df.rename(columns={diagnosis_col: 'Disease_Category'}, inplace=True)

print(f"✓ Epidemic data created: {epidemic_df.shape}")
print(f"✓ Columns: {epidemic_df.columns.tolist()}")
print(f"\nSample:")
print(epidemic_df.head(10))

# ⭐ DO NOT CREATE EPIDEMIC COLUMN HERE! ⭐
# We'll create it AFTER splitting in Cell #34
```

```python
#34 - PROPER FEATURE ENGINEERING (NO DATA LEAKAGE!)

print("="*70)
print("EPIDEMIC FEATURE ENGINEERING - PREVENTING DATA LEAKAGE")
print("="*70)

# The diagnosis column in epidemic_df is called 'Disease_Category'
disease_col = 'Disease_Category'

print(f"✓ Using disease column: {disease_col}")
print(f"✓ epidemic_df shape: {epidemic_df.shape}")
print(f"✓ Columns before processing: {epidemic_df.columns.tolist()}")

# ⭐ STEP 1: Create time-based features (these are safe)
epidemic_df['month_sin'] = np.sin(2 * np.pi * epidemic_df['month'] / 12)
epidemic_df['month_cos'] = np.cos(2 * np.pi * epidemic_df['month'] / 12)

# ⭐ STEP 2: Sort and create lag features (by disease group)
epidemic_df = epidemic_df.sort_values(['Disease_Category', 'month']).reset_index(drop=True)
epidemic_df['cases_lag_1'] = epidemic_df.groupby(disease_col)['cases'].shift(1)
epidemic_df['cases_lag_2'] = epidemic_df.groupby(disease_col)['cases'].shift(2)

# ⭐ STEP 3: Rolling features
epidemic_df['cases_rolling_mean_3'] = epidemic_df.groupby(disease_col)['cases'].rolling(3, min_periods=1).mean().reset_index(0, drop=True)
epidemic_df['cases_rolling_std_3'] = epidemic_df.groupby(disease_col)['cases'].rolling(3, min_periods=1).std().reset_index(0, drop=True)

# Fill remaining NaN with 0
epidemic_df = epidemic_df.fillna(0)

# ⭐ STEP 4: Encode disease type
from sklearn.preprocessing import LabelEncoder
disease_encoder = LabelEncoder()
epidemic_df['Disease_Encoded'] = disease_encoder.fit_transform(epidemic_df[disease_col])

print(f"\n📊 Disease Categories:")
for i, disease in enumerate(disease_encoder.classes_):
    count = (epidemic_df['Disease_Encoded'] == i).sum()
    print(f"   {i}: {disease} ({count} data points)")

# ⭐ STEP 5: SPLIT DATA FIRST (before creating target!)
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Define features BEFORE splitting
# Define features BEFORE splitting
feature_cols = ['month', 'cases_lag_1', 'cases_lag_2',
                'cases_rolling_mean_3', 'cases_rolling_std_3',
                'month_sin', 'month_cos', 'Disease_Encoded']

X = epidemic_df[feature_cols].copy()

# SPLIT FIRST - 70% train, 30% test
X_train_raw, X_test_raw = train_test_split(
    X,
    test_size=0.3,
    random_state=42,
    shuffle=True
)

print(f"\n✅ Data split:")
print(f"   Train size: {len(X_train_raw)}")
print(f"   Test size: {len(X_test_raw)}")

# Get corresponding epidemic_df rows
train_indices = X_train_raw.index
test_indices = X_test_raw.index

epidemic_train = epidemic_df.loc[train_indices].copy()
epidemic_test = epidemic_df.loc[test_indices].copy()

# ⭐ STEP 6: Create threshold using ONLY TRAINING DATA
train_threshold = epidemic_train["cases"].quantile(0.75)
print(f"\n✅ Training threshold (75th percentile): {train_threshold:.2f} cases")

# Create target variable separately for train and test
epidemic_train["Epidemic"] = (epidemic_train["cases"] >= train_threshold).astype(int)
epidemic_test["Epidemic"] = (epidemic_test["cases"] >= train_threshold).astype(int)

print(f"\n📊 Training set:")
print(f"   Total: {len(epidemic_train)}")
print(f"   Epidemic (1): {epidemic_train['Epidemic'].sum()}")
print(f"   Normal (0): {(epidemic_train['Epidemic']==0).sum()}")
print(f"   Balance: {epidemic_train['Epidemic'].mean():.1%} epidemic")

print(f"\n📊 Test set:")
print(f"   Total: {len(epidemic_test)}")
print(f"   Epidemic (1): {epidemic_test['Epidemic'].sum()}")
print(f"   Normal (0): {(epidemic_test['Epidemic']==0).sum()}")
print(f"   Balance: {epidemic_test['Epidemic'].mean():.1%} epidemic")

# Extract features and target
X_train_epi = epidemic_train[feature_cols].copy()
y_train_epi = epidemic_train["Epidemic"].copy()
X_test_epi = epidemic_test[feature_cols].copy()
y_test_epi = epidemic_test["Epidemic"].copy()

# ⭐ STEP 7: Scale features (fit on train, transform test)
scaler = StandardScaler()
X_train_epi_scaled = scaler.fit_transform(X_train_epi)
X_test_epi_scaled = scaler.transform(X_test_epi)

# Convert back to DataFrames with proper indices
X_train_epi = pd.DataFrame(X_train_epi_scaled, columns=feature_cols, index=train_indices)
X_test_epi = pd.DataFrame(X_test_epi_scaled, columns=feature_cols, index=test_indices)

# ⭐ STEP 8: Store the split epidemic_df for later use in alerts
# We'll use these to map predictions back to disease names
epidemic_train_for_output = epidemic_train.copy()
epidemic_test_for_output = epidemic_test.copy()

print("\n✅ DATA PREPARED WITHOUT LEAKAGE!")
print(f"   X_train_epi shape: {X_train_epi.shape}")
print(f"   y_train_epi shape: {y_train_epi.shape}")
print(f"   X_test_epi shape: {X_test_epi.shape}")
print(f"   y_test_epi shape: {y_test_epi.shape}")

print("\n⚠️  CRITICAL: Each model will now be trained on DIFFERENT data!")
print("   This will fix the 'all models same confusion matrix' issue.")
```

```python
#78
def predict_patient_risk(patient_data, model, scaler, le_target,
                        feature_columns, preventive_measures, risk_based_measures):
    """Predict disease risk and provide preventive measures"""

    # Prepare features
    X_patient = patient_data[feature_columns].copy()
    X_patient_scaled = scaler.transform(X_patient)

    # Predict
    prediction = model.predict(X_patient_scaled)[0]
    prediction_proba = model.predict_proba(X_patient_scaled)[0]

    # Get disease category
    predicted_disease = le_target.inverse_transform([prediction])[0]
    confidence = prediction_proba[prediction] * 100

    # Get top 3 risks
    top_3_indices = np.argsort(prediction_proba)[-3:][::-1]
    top_3_diseases = le_target.inverse_transform(top_3_indices)
    top_3_probas = prediction_proba[top_3_indices] * 100

    # Identify active risk factors
    active_risks = []
    if patient_data['Diabetes_Risk'].values[0] == 1:
        active_risks.extend(risk_based_measures.get('High_Diabetes_Risk', []))
    if patient_data['Kidney_Dysfunction_Risk'].values[0] == 1:
        active_risks.extend(risk_based_measures.get('High_Kidney_Risk', []))
    if patient_data['Hypertension_Risk'].values[0] == 1:
        active_risks.extend(risk_based_measures.get('High_BP_Risk', []))
    if patient_data['Anemia_Risk'].values[0] == 1:
        active_risks.extend(risk_based_measures.get('Anemia_Risk', []))

    # Get preventive measures
    disease_measures = preventive_measures.get(predicted_disease, preventive_measures.get('Other_Condition', {}))

    # Risk level
    risk_level = 'HIGH' if confidence > 70 else 'MODERATE' if confidence > 50 else 'LOW'
    timeline = '1-3 months' if confidence > 70 else '3-6 months' if confidence > 50 else '6-12 months'

    return {
        'disease': predicted_disease,
        'confidence': confidence,
        'risk_level': risk_level,
        'timeline': timeline,
        'top_risks': list(zip(top_3_diseases, top_3_probas)),
        'active_risks': active_risks,
        'preventive_measures': disease_measures
    }

print("✓ Prediction function created")
```
